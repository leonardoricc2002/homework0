```python
%%writefile somma.cu
#include <stdio.h>

//void sommaCPU(int *a, int*b , int *c ,int N){
//    for(int i =0; i<N; i++)
//    c[i]=a[i]+b[i];
//}
__global__ void sommaGPU(int*d_a ,int*d_b ,int*d_c ,int N)
{
      int tid = threadIdx.x;
      d_c[tid]=d_a[tid]+d_b[tid];
}


int main(){

  //alloco su CPU
    const int N=5;
    int*a = (int*)malloc(N*sizeof(int));
    int*b = (int*)malloc(N*sizeof(int));
    int*c = (int*)malloc(N*sizeof(int));

   //alloco su GPU
    int* d_a;
    int* d_b;
    int* d_c;
    cudaMalloc(&d_a,N*sizeof(int));// se restituisce 0 è ok!
    cudaMalloc(&d_b,N*sizeof(int));
    cudaMalloc(&d_c,N*sizeof(int));

    for(int i =0; i<N; i++){
      a[i]=i;
      b[i]=2*i;
    }
  //trasporto da HOST A DEVICE ,CPU-->GPU
    cudaMemcpy(d_a,a,N*sizeof(int),cudaMemcpyHostToDevice); //anche queste restituisce numero
    cudaMemcpy(d_b,b,N*sizeof(int),cudaMemcpyHostToDevice);

    //sommaCPU(a,b,c,N);
   // for(int i =0; i<N; i++)
    // printf("Elemento %d = %d\n", i,c[i]);

    sommaGPU<<<1,5>>>(d_a,d_b,d_c,N);
    cudaMemcpy(c,d_c,N*sizeof(int),cudaMemcpyDeviceToHost);//Per dare i dati alla CPU

   for(int i =0; i<N; i++){
       printf("Elemento %d = %d\n", i,c[i]);
   }

}

```
